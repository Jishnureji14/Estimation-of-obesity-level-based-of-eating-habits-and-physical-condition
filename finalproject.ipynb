```python
age_bmi = df.groupby('Age_Group')['BMI'].mean().reset_index()
```

### **1. Age Distribution**

**What is the age distribution of individuals in the dataset ?**





```python
sns.histplot(df['Age'])
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

***The age distribution shows that most individuals fall within the young to middle-adult age group. The spread indicates that obesity-related lifestyle patterns are being observed mainly among adults rather than children or elderly individuals, making the dataset suitable for lifestyle-based obesity analysis.***

### **2. Weight Distribution**

**How is body weight distributed among individuals ?**

```python
sns.histplot(df['Weight'])
plt.title("Weight Distribution")
plt.xlabel("Weight")
plt.ylabel("Frequency")
plt.show()
```

***The weight distribution is widely spread, showing clear variation among individuals. This variation supports the presence of multiple obesity categories and indicates that body weight plays a major role in distinguishing between different obesity levels.***

### **3. Obesity Level Distribution**

**How are individual distributed across obesity levels ?**

```python
sns.countplot(x='NObeyesdad', data=df)
plt.title("Obesity Level Distribution")
plt.xticks(rotation=45)
plt.show()
```

**Some obesity categories have more individuals than others, indicating class imbalance.**

### **4. Transportation Distribution**

**What is the distribution of mode of transportation ?**

```python
transport_counts = df['MTRANS'].value_counts()

transport_counts.sort_values().plot(kind='barh')
plt.title("Distribution of Mode of Transportation")
plt.xlabel("Number of Individuals")
plt.ylabel("Mode of Transportation")
plt.show()
```

**Most individuals use passive transportation like cars and buses, while fewer rely on active modes such as walking or cycling. This suggests limited daily physical movement, which may contribute to a sedentary lifestyle.**

### **5. Gender Distribution**

**What is the gender distribution in the dataset ?**

```python
df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Gender Distribution")
plt.ylabel("")
plt.show()
```

***The dataset contains representation from both genders with a fairly balanced distribution.The gender distribution appears relatively balanced, meaning both male and female lifestyle patterns are represented. This balance improves the reliability of conclusions drawn from behavioral comparisons.***

### **6. Height Vs Weight**

**What is the relationship between Height and Weight ?**

```python
sns.scatterplot(x='Height', y='Weight', hue='NObeyesdad', data=df)
plt.title("Height vs Weight")
plt.show()
```

***Weight increases with height, but obesity categories form distinct clusters.The scatter plot shows a positive relationship between height and weight. However, clusters of obesity categories form at higher weight ranges, indicating that weight contributes more strongly to obesity classification than height alone.***

### **7. Physical Activity Vs Obesity Level**

**How does physical activity vary across obesity levels ?**

```python
sns.boxplot(x='NObeyesdad', y='FAF', data=df)
plt.title("Physical Activity vs Obesity Level")
plt.xticks(rotation=45)
plt.show()
```

***This figure shows that individuals in higher obesity categories generally have lower physical activity levels. Normal-weight individuals tend to have higher activity frequency, suggesting that regular exercise plays an important role in maintaining healthy weight.***

### **8. Technology Usage Vs Obesity Level**

**How does technology usage relate to obesity levels ?**

```python
sns.boxplot(x='NObeyesdad', y='TUE', data=df)
plt.title("Technology Usage vs Obesity Level")
plt.xticks(rotation=45)
plt.show()
```

***Higher obesity levels are associated with increased technology usage time. This suggests that sedentary behaviors, such as long screen time, may contribute to weight gain and reduced physical activity.***

### **9. Water Intake Vs Obesity Levels**

**How does water intake vary across obesity levels ?**

```python
sns.boxplot(x='NObeyesdad', y='CH2O', data=df)
plt.title("Water Intake vs Obesity Level")
plt.xticks(rotation=45)
plt.show()
```

***The distribution shows that individuals with lower water consumption appear more frequently in higher obesity groups. While the relationship is moderate, it suggests hydration habits may be linked with overall lifestyle health.***

### **10. Average Physical Activity by Gender**

**How does physical activity differ between genders ?**


```python
df.groupby('Gender')['FAF'].mean().plot(kind='bar')
plt.title("Average Physical Activity by Gender")
plt.ylabel("Mean FAF")
plt.show()
```

***This figure compares the average physical activity frequency between males and females. The analysis shows a slight difference in activity levels between genders, indicating that lifestyle patterns related to exercise may vary.***

### **11. Correlation Heatmap of Lifestyle Factors**

**What relationships exist among lifestyle-related factors ?**



```python
lifestyle = ['FCVC', 'NCP', 'CH2O', 'FAF', 'TUE']

plt.figure(figsize=(8,5))
sns.heatmap(df[lifestyle].corr(), annot=True, cmap='YlGnBu')
plt.title("Correlation Between Lifestyle Factors")
plt.show()
```

***Lifestyle behaviors in this dataset tend to act independently rather than as strongly connected habits. This means each factor — diet, physical activity, hydration, and screen time — may influence obesity separately rather than as a single combined behavior pattern.***

### **12. BMI Vs Age**

**How does BMI vary with Age ?**

```python
sns.lineplot(data=age_bmi, x='Age_Group', y='BMI', marker='o')
plt.title("Average BMI Across Age Groups")
plt.xlabel("Age Group")
plt.ylabel("Average BMI")
plt.show()
```

***The plot shows that average BMI increases from younger age groups and peaks around the 40–50 age group, indicating a higher risk of overweight and obesity during middle age. After this stage, the average BMI shows a slight decline in older age groups (50+), which may be influenced by lifestyle changes, health conditions, or reduced body mass with aging. Overall, the trend suggests that middle-aged individuals are more prone to higher BMI levels compared to younger individuals.***

The ultimate goal is to understand which lifestyle behaviors contribute most to obesity and how healthier habits can help reduce obesity risk.*

### ***Data Preparation Summary***

*The original dataset consisted of 2111 records containing a mix of numerical and categorical features related to demographic characteristics, lifestyle habits, and physical condition indicators. ### ***Exploratory Data Analysis Overview***

*The exploratory data analysis (EDA) phase was conducted to identify patterns, trends, and relationships among demographic, physical, and lifestyle variables associated with obesity levels. Various univariate, bivariate, and multivariate visualizations were used to understand data distributions and interactions between features. The analysis revealed that weight, BMI, and physical activity levels show strong variation across obesity categories, while lifestyle factors such as sedentary behavior, dietary habits, and transportation methods also contribute to observable differences. Correlation analysis indicated that most lifestyle variables are weakly related, suggesting that obesity is influenced by multiple independent behaviors rather than a single dominant factor. Overall, the EDA provided meaningful insights into how daily habits and physical attributes collectively impact obesity levels.*

### **Age Vs Obesity**

*The age distribution shows that most individuals in the dataset fall within young to middle adulthood. There is a noticeable spread across different age groups, allowing comparison of obesity levels throughout adulthood. BMI and weight-related measures tend to show gradual variation as age increases.*

*The analysis suggests that obesity levels tend to increase slightly with age. This may be due to reduced physical activity, slower metabolism, and lifestyle changes over time. Although age itself is not the only cause of obesity, it acts as an influencing factor that can increase obesity risk when combined with unhealthy habits.*

### **Physical Activity Vs Obesity**

*The distribution of physical activity levels shows clear variation across individuals in the dataset. Some participants report regular physical activity, while others show very low activity levels. When compared across obesity categories, individuals with lower activity frequency appear more concentrated in overweight and obese groups.*

*The analysis indicates a strong relationship between low physical activity and higher obesity levels. Physical inactivity reduces calorie expenditure, which can lead to weight gain over time. This suggests that regular exercise plays a significant role in maintaining a healthy weight and lowering the risk of obesity.*

### **Technology Vs Obesity**

*The analysis of technology usage shows varying levels of daily screen time among individuals. A noticeable portion of participants report spending longer hours using electronic devices such as phones, computers, or television. Higher screen time appears more frequently in individuals belonging to overweight and obese categories.*

*Increased technology usage is associated with a sedentary lifestyle, which reduces physical movement and calorie expenditure. Prolonged screen time often replaces active behaviors and may also be linked with unhealthy snacking habits. This suggests that excessive technology use contributes indirectly to higher obesity risk by promoting inactivity.*

### **Transportation Vs Obesity**

*The distribution of transportation modes shows that individuals use a mix of active and passive travel methods. Passive modes such as cars, buses, and motorbikes are more common overall, while fewer individuals rely on walking or cycling. Higher obesity categories tend to have a greater proportion of people using passive transportation.*

*Transportation choice directly influences daily physical activity levels. Individuals who rely on passive transport expend fewer calories in their routine, which may contribute to gradual weight gain. In contrast, active transportation like walking or cycling supports regular movement and is associated with healthier weight levels. This suggests that daily mobility habits play an important role in obesity risk.*

### **Water Intake Vs Obesity**

*The analysis shows variation in daily water consumption among individuals. Some participants report higher water intake, while others consume less. When compared across obesity levels, individuals in normal-weight categories tend to appear slightly more in higher water intake groups.*

*Adequate water intake supports healthy metabolism and can help regulate appetite. Lower water consumption may sometimes be associated with unhealthy eating habits or reduced overall health awareness. Although the relationship is not extremely strong, proper hydration appears to be a supportive factor in maintaining a healthy weight and preventing obesity.*

### **Dietary Habits Vs Obesity**

*The analysis of dietary habits shows clear differences in eating patterns among individuals. Frequent intake of high-calorie foods is more commonly observed in overweight and obese categories. In contrast, individuals with healthier and more balanced eating patterns appear more often in normal-weight groups.*

*Diet plays a major role in body weight regulation. Regular consumption of high-calorie, low-nutrient foods increases energy intake, which can lead to fat accumulation over time. On the other hand, balanced and nutritious eating habits support better metabolism and weight control. This suggests that unhealthy eating behaviors are a significant contributor to obesity risk.*

### **Recommendations for Reducing Obesity**

**Maintain adequate daily water intake**
